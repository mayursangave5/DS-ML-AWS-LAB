Problem: Predict house price from house area.

Available data: Historical house area and price.

Input: House area

What we want: House price

Question: Can we learn the relationship between area and price from historical examples and use it for a new house?

```python
# x= [[500], [700], [800], [1000], [1200], [1500], [1800], [2000]]
x= [[500], [700], [900], [1100], [1300], [1500]]
# y= [30, 42, 48, 60, 70, 85, 100, 115]
y= [100, 20, 90, 30, 120, 40]
print(x)
print(y)
```

Each training example needs a corresponding target value. Therefore, the number of samples in X and y must match.

```python
import matplotlib.pyplot as plt

plt.scatter(x,y)
plt.xlabel("Area")
plt.title("House area vs prise")
plt.ylabel("Prise in lakhs")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x,y)
```

```python
prediction=model.predict([[1200]])
print(prediction)
```

Note:-
Building a model is only one part of an ML project. We must determine whether the model's predictions are actually useful.

```python
print(model.coef_)
print(model.intercept_)
# The model learned numbers that describe the relationship between area and price.
```

```python
print(model.predict([[500]]))
print(model.predict([[1000]]))
print(model.predict([[1200]]))
print(model.predict([[2000]]))
print(model.predict([[3000]]))
```

A machine-learning model learns from the data you give it. 
If your data is poor, biased, incorrect, or unrepresentative, the model can learn a poor relationship.
